# 1. Знакомимся с игрой — 0–10 мин

### Как работает «Змейка»?

Представьте поле как тетрадь в клетку:

1. Змейка состоит из **головы** и **частей тела**.
2. Каждый ход голова перемещается **на одну клетку**.
3. Остальные части тела следуют за головой.
4. Игрок нажимает стрелки, чтобы выбрать направление.
5. На втором занятии добавим еду, рост и очки.

### Важные слова

| Слово | Простое объяснение |
|---|---|
| **Пиксель** | Маленькая точка на экране |
| **Клетка** | Наш игровой квадратик размером 20 × 20 пикселей |
| **Координаты X, Y** | Адрес клетки: столбец и строка |
| **Игровой цикл** | Постоянное повторение: проверить клавиши → обновить игру → нарисовать кадр |

# 3. Наше первое игровое окно — 20–35 мин

Создадим окно **600 × 400 пикселей**. Цвет задаётся тремя числами: **(красный, зелёный, синий)**. Каждое число — от 0 до 255.

Примеры:
- `(0, 0, 0)` — чёрный;
- `(255, 255, 255)` — белый;
- `(40, 60, 80)` — тёмно-синий.

### Запустите код

Откроется отдельное окно. **Закройте его крестиком**, чтобы ячейка завершилась.

````python
import pygame

pygame.init()  # Запускаем Pygame

WIDTH = 600
HEIGHT = 400

screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Моя первая игра")
clock = pygame.time.Clock()

running = True
while running:
    # Проверяем, не закрыли ли окно
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # Закрашиваем окно и показываем кадр
    screen.fill((40, 60, 80))
    pygame.display.flip()
    clock.tick(30)  # Не больше 30 кадров в секунду

pygame.quit()
````

### Разбираем главные команды

| Команда | Что делает |
|---|---|
| `pygame.init()` | Подготавливает библиотеку к работе |
| `set_mode((600, 400))` | Создаёт окно шириной 600 и высотой 400 |
| `set_caption(...)` | Меняет надпись сверху окна |
| `while running:` | Повторяет действия, пока окно не закрыто |
| `pygame.event.get()` | Получает события: нажатия клавиш, закрытие окна и другие |
| `screen.fill(...)` | Заливает фон одним цветом |
| `pygame.display.flip()` | Показывает нарисованный кадр |
| `clock.tick(30)` | Ограничивает скорость обновления окна |
| `pygame.quit()` | Завершает работу Pygame |

# 4. Координаты и рисование — 35–45 мин

В Pygame начало координат **(0, 0)** находится **в левом верхнем углу**:

```text
(0,0) ─────────────────────→ X
  │
  │    (100, 80) ■
  │
  ▼ Y
```

- Увеличиваем **X** — идём **вправо**.
- Увеличиваем **Y** — идём **вниз**.
- Змейку будем рисовать из квадратов по **20 пикселей**.

**Важно:** сначала рисуем фон, затем фигуры, и только после этого вызываем `flip()`.

````python
import pygame
pygame.init()

WIDTH = 600
HEIGHT = 400
CELL = 20  # Размер одного квадратика

screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Рисуем фигуры")
clock = pygame.time.Clock()

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    screen.fill((25, 35, 55))

    # Зелёный квадрат: x, y, ширина, высота
    pygame.draw.rect(screen, (70, 210, 100), (100, 120, CELL, CELL))

    # Красный круг: центр x/y и радиус
    pygame.draw.circle(screen, (240, 70, 70), (280, 180), 10)

    pygame.display.flip()
    clock.tick(30)

pygame.quit()
````

Один квадрат — это только голова. А как нарисовать тело?

Используем **список координат**, в котором каждый маленький список — один сегмент:

```python

```

Здесь `[10, 7]` — **голова**: столбец 10, строка 7. Остальные два элемента — тело. Номер клетки умножаем на 20, чтобы получить координату **в пикселях**.

Например: столбец `10` → `10 * 20 = 200` пикселей.

````python
# Мини-опыт со списками: окно не открывается
snake = [[10, 7], [9, 7], [8, 7]]

print("Вся змейка:", snake)
print("Голова:", snake[0])
print("X головы в пикселях:", snake[0][0] * 20)
print("Y головы в пикселях:", snake[0][1] * 20)
````

Теперь пройдём по **всем сегментам** циклом `for` и нарисуем каждый прямоугольником. В отличие от прошлого примера, координаты здесь записаны **в клетках**, а не в пикселях.

````python
import pygame
pygame.init()

CELL = 20
WIDTH = 600
HEIGHT = 400
snake = [[10, 7], [9, 7], [8, 7]]

screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Моя змейка")
clock = pygame.time.Clock()

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    screen.fill((25, 35, 55))

    for part in snake:
        x = part[0] * CELL  # Номер столбца → пиксели
        y = part[1] * CELL  # Номер строки → пиксели
        pygame.draw.rect(screen, (70, 210, 100), (x, y, CELL, CELL))

    pygame.display.flip()
    clock.tick(30)

pygame.quit()
````

# 6. Змейка оживает: движение и стрелки — 55–70 мин

Чтобы змейка двигалась, на каждом шаге нужно:

1. Узнать, где находится **голова**.
2. Создать **новую голову на одну клетку впереди**.
3. Добавить её **в начало списка** через `insert(0, ...)`.
4. Удалить **последний сегмент** через `pop()`.

Так длина не меняется, зато змейка «ползёт».

### Проверьте принцип без игрового окна

````python
snake = [[10, 7], [9, 7], [8, 7]]
print("До шага:", snake)

new_head = [snake[0][0] + 1, snake[0][1]]  # Один шаг вправо
snake.insert(0, new_head)                # Добавили новую голову
snake.pop()                              # Убрали хвост

print("После шага:", snake)
# Ожидаем: [[11, 7], [10, 7], [9, 7]]
````

### Управление

Переменная `direction` хранит слово **`"RIGHT"`**, **`"LEFT"`**, **`"UP"`** или **`"DOWN"`**. При нажатии стрелки мы меняем это слово. На каждом шаге проверяем его и сдвигаем новую голову.

Выражение `direction != "DOWN"` означает «направление **не равно** `DOWN`». Это нужно, чтобы змейка не разворачивалась сразу на 180° и не двигалась в собственное тело.

**Обратите внимание:** мы используем `clock.tick(5)`, чтобы змейка делала примерно **5 шагов в секунду**, а не двигалась слишком быстро.

**Запустите следующую версию и попробуйте все стрелки.** Пока змейка может выйти за поле — на следующем этапе исправим.

````python
import pygame
pygame.init()

CELL = 20
WIDTH = 600
HEIGHT = 400
snake = [[10, 7], [9, 7], [8, 7]]
direction = "RIGHT"

screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Управляем змейкой")
clock = pygame.time.Clock()

running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        elif event.type == pygame.KEYDOWN:
            if event.key == pygame.K_UP and direction != "DOWN":
                direction = "UP"
            elif event.key == pygame.K_DOWN and direction != "UP":
                direction = "DOWN"
            elif event.key == pygame.K_LEFT and direction != "RIGHT":
                direction = "LEFT"
            elif event.key == pygame.K_RIGHT and direction != "LEFT":
                direction = "RIGHT"

    # Берём копию головы и сдвигаем её на одну клетку
    new_head = snake[0].copy()
    if direction == "RIGHT":
        new_head[0] += 1
    elif direction == "LEFT":
        new_head[0] -= 1
    elif direction == "UP":
        new_head[1] -= 1
    elif direction == "DOWN":
        new_head[1] += 1

    snake.insert(0, new_head)
    snake.pop()

    screen.fill((25, 35, 55))
    for part in snake:
        x = part[0] * CELL
        y = part[1] * CELL
        pygame.draw.rect(screen, (70, 210, 100), (x, y, CELL, CELL))

    pygame.display.flip()
    clock.tick(5)

pygame.quit()
````

# 7. Не выходим за края: итоговая версия — 70–80 мин

Сейчас змейка умеет двигаться, но может исчезнуть за экраном. Добавим проверку границ.

При размере клетки 20 пикселей в окне 600 × 400 помещается:

- **30 столбцов:** `600 // 20 = 30`;
- **20 строк:** `400 // 20 = 20`.

Допустимые координаты: X — от **0 до 29**, Y — от **0 до 19**.

Проверка:

```python
0 <= new_head[0] < COLS and 0 <= new_head[1] < ROWS
```

Если новая голова находится внутри поля, змейка движется. Иначе змейка **останавливается**, а окно остаётся открытым — его можно закрыть крестиком. Настоящий экран проигрыша добавим на втором уроке.

````python
import pygame

pygame.init()

# Размер игрового поля
CELL = 20
COLS = 30
ROWS = 20
WIDTH = COLS * CELL
HEIGHT = ROWS * CELL

# Цвета
BACKGROUND = (25, 35, 55)
SNAKE_COLOR = (70, 210, 100)
HEAD_COLOR = (120, 250, 150)

# Начальное положение змейки
snake = [[10, 7], [9, 7], [8, 7]]
direction = "RIGHT"
moving = True

screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Змейка — урок 1")
clock = pygame.time.Clock()

running = True
while running:
    # 1. Читаем события
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        elif event.type == pygame.KEYDOWN and moving:
            if event.key == pygame.K_UP and direction != "DOWN":
                direction = "UP"
            elif event.key == pygame.K_DOWN and direction != "UP":
                direction = "DOWN"
            elif event.key == pygame.K_LEFT and direction != "RIGHT":
                direction = "LEFT"
            elif event.key == pygame.K_RIGHT and direction != "LEFT":
                direction = "RIGHT"

    # 2. Передвигаем змейку, пока она не столкнулась со стеной
    if moving:
        new_head = snake[0].copy()
        if direction == "RIGHT":
            new_head[0] += 1
        elif direction == "LEFT":
            new_head[0] -= 1
        elif direction == "UP":
            new_head[1] -= 1
        elif direction == "DOWN":
            new_head[1] += 1

        # Проверяем, осталась ли голова внутри поля
        if 0 <= new_head[0] < COLS and 0 <= new_head[1] < ROWS:
            snake.insert(0, new_head)
            snake.pop()
        else:
            moving = False
            pygame.display.set_caption("Змейка у стены! Закройте окно")

    # 3. Рисуем поле и тело змейки
    screen.fill(BACKGROUND)
    for part in snake:
        x = part[0] * CELL
        y = part[1] * CELL
        pygame.draw.rect(screen, SNAKE_COLOR, (x, y, CELL, CELL))

    # Рисуем голову другим цветом
    head_x = snake[0][0] * CELL
    head_y = snake[0][1] * CELL
    pygame.draw.rect(screen, HEAD_COLOR, (head_x, head_y, CELL, CELL))

    pygame.display.flip()
    clock.tick(5)  # 5 игровых шагов в секунду

pygame.quit()
````

## 8.1. Что такое класс — 5 минут

**Класс** — это инструкция, по которой мы можем создать объект. Например, класс `Snake` описывает, что змейка **знает** и что она **умеет**.

| Понятие | На нашем примере |
|---|---|
| `class Snake:` | Описание змейки |
| `snake = Snake()` | Создали змейку — **объект** класса |
| `__init__` | Настраивает объект при создании |
| `self.body` | Данные **этой** змейки: список частей тела |
| `def move(self):` | **Метод** — действие, которое умеет выполнять змейка |
| `snake.move()` | Попросили объект сделать шаг |

**Важно:** слово `self` означает «этот конкретный объект». Его не нужно передавать при вызове `snake.move()` — Python делает это сам.

Ниже — маленький опыт **без игрового окна**, чтобы увидеть идею класса.

````python
class Player:
    def __init__(self):
        self.name = "Змейка"
        self.steps = 0

    def move(self):
        self.steps += 1
        print(self.name, "сделала шаг. Всего:", self.steps)

player = Player()  # Создали объект
player.move()      # Вызвали метод
player.move()

# Попробуйте изменить имя на своё и сделать ещё один шаг.
````

## 8.2. Переносим змейку в класс `Snake` — 10 минут

Посмотрите на итоговый код из раздела 7. Там мы использовали переменные `snake`, `direction`, `moving` и команды изменения направления, движения и рисования. Теперь помещаем **эти же действия** внутрь одного класса:

| Раньше в большом коде | Теперь внутри класса `Snake` |
|---|---|
| `snake = [[10, 7], ...]` | `self.body = [[10, 7], ...]` |
| `direction = "RIGHT"` | `self.direction = "RIGHT"` |
| `moving = True` | `self.moving = True` |
| Проверка `event.key` | `turn(key)` |
| Создание `new_head`, `insert`, `pop` | `move()` |
| Цикл `for part in snake` | `draw(screen)` |

**Заметьте:** внутри класса к данным добавляется `self.`. Например, `snake[0]` превращается в `self.body[0]`.

Запустите следующий блок. Он **не открывает окно**: мы проверяем, что объект-змейка двигается правильно. Метод `draw()` пригодится позже, когда снова создадим окно.

````python
import pygame

CELL = 20
COLS = 30
ROWS = 20
SNAKE_COLOR = (70, 210, 100)
HEAD_COLOR = (120, 250, 150)


class Snake:
    def __init__(self):
        # Список клеток: первая — голова
        self.body = [[10, 7], [9, 7], [8, 7]]
        self.direction = "RIGHT"
        self.moving = True

    def turn(self, key):
        # Не даём развернуться в обратную сторону
        if not self.moving:
            return

        if key == pygame.K_UP and self.direction != "DOWN":
            self.direction = "UP"
        elif key == pygame.K_DOWN and self.direction != "UP":
            self.direction = "DOWN"
        elif key == pygame.K_LEFT and self.direction != "RIGHT":
            self.direction = "LEFT"
        elif key == pygame.K_RIGHT and self.direction != "LEFT":
            self.direction = "RIGHT"

    def move(self):
        if not self.moving:
            return

        # Начинаем с копии головы
        new_head = self.body[0].copy()
        if self.direction == "RIGHT":
            new_head[0] += 1
        elif self.direction == "LEFT":
            new_head[0] -= 1
        elif self.direction == "UP":
            new_head[1] -= 1
        elif self.direction == "DOWN":
            new_head[1] += 1

        # Перемещаемся, только если клетка внутри поля
        if 0 <= new_head[0] < COLS and 0 <= new_head[1] < ROWS:
            self.body.insert(0, new_head)
            self.body.pop()
        else:
            self.moving = False

    def draw(self, screen):
        for part in self.body:
            x = part[0] * CELL
            y = part[1] * CELL
            pygame.draw.rect(screen, SNAKE_COLOR, (x, y, CELL, CELL))

        # Голова выделена другим цветом
        head_x = self.body[0][0] * CELL
        head_y = self.body[0][1] * CELL
        pygame.draw.rect(screen, HEAD_COLOR, (head_x, head_y, CELL, CELL))


snake = Snake()
print("До движения:", snake.body)
snake.move()
print("После движения:", snake.body)
print("Направление:", snake.direction)
````

## 8.3. Создаём класс `Game` — 10 минут

Теперь нужно вернуть **игровое окно**. В исходной программе окно и игровой цикл уже были. Перенесём их в `Game`.

- `__init__()` — открывает окно, создаёт таймер и объект `Snake`.
- `run()` — повторяет три знакомых шага: **события → движение → рисование**.
- `self.snake.turn(event.key)` — сообщает змейке, какую кнопку нажал игрок.
- `self.snake.move()` — просит змейку сделать следующий шаг.
- `self.snake.draw(self.screen)` — просит змейку нарисовать себя.

`Game` не должен знать, *как именно* переставляются части тела змейки: этим занимается `Snake`.

> **Порядок запуска:** сначала выполните предыдущую ячейку с объявлением `class Snake`, затем ячейку ниже. Закройте окно крестиком, прежде чем продолжать.

````python
WIDTH = COLS * CELL
HEIGHT = ROWS * CELL
BACKGROUND = (25, 35, 55)
SPEED = 5


class Game:
    def __init__(self):
        pygame.init()
        self.screen = pygame.display.set_mode((WIDTH, HEIGHT))
        pygame.display.set_caption("Змейка — урок 1")
        self.clock = pygame.time.Clock()
        self.snake = Snake()  # Объект нашего класса

    def run(self):
        running = True
        while running:
            for event in pygame.event.get():
                if event.type == pygame.QUIT:
                    running = False
                elif event.type == pygame.KEYDOWN:
                    self.snake.turn(event.key)

            self.snake.move()
            if not self.snake.moving:
                pygame.display.set_caption("Змейка у стены! Закройте окно")

            self.screen.fill(BACKGROUND)
            self.snake.draw(self.screen)
            pygame.display.flip()
            self.clock.tick(SPEED)

        pygame.quit()


# Теперь всю игру можно запустить двумя строками:
game = Game()
game.run()
````

## 9.3. Файл `snake.py` — только змейка

Переносим сюда знакомый класс `Snake`. Первые две строки — **импорт**: мы берём Pygame и настройки из других файлов.

`from settings import CELL, COLS, ...` — это просто «возьми эти значения из файла `settings.py`».

````python
%%writefile snake_oop/snake.py
import pygame
from settings import CELL, COLS, ROWS, SNAKE_COLOR, HEAD_COLOR


class Snake:
    def __init__(self):
        # Список клеток: первая — голова
        self.body = [[10, 7], [9, 7], [8, 7]]
        self.direction = "RIGHT"
        self.moving = True

    def turn(self, key):
        # Не даём развернуться в обратную сторону
        if not self.moving:
            return

        if key == pygame.K_UP and self.direction != "DOWN":
            self.direction = "UP"
        elif key == pygame.K_DOWN and self.direction != "UP":
            self.direction = "DOWN"
        elif key == pygame.K_LEFT and self.direction != "RIGHT":
            self.direction = "LEFT"
        elif key == pygame.K_RIGHT and self.direction != "LEFT":
            self.direction = "RIGHT"

    def move(self):
        if not self.moving:
            return

        # Начинаем с копии головы
        new_head = self.body[0].copy()
        if self.direction == "RIGHT":
            new_head[0] += 1
        elif self.direction == "LEFT":
            new_head[0] -= 1
        elif self.direction == "UP":
            new_head[1] -= 1
        elif self.direction == "DOWN":
            new_head[1] += 1

        # Перемещаемся, только если клетка внутри поля
        if 0 <= new_head[0] < COLS and 0 <= new_head[1] < ROWS:
            self.body.insert(0, new_head)
            self.body.pop()
        else:
            self.moving = False

    def draw(self, screen):
        for part in self.body:
            x = part[0] * CELL
            y = part[1] * CELL
            pygame.draw.rect(screen, SNAKE_COLOR, (x, y, CELL, CELL))

        # Голова выделена другим цветом
        head_x = self.body[0][0] * CELL
        head_y = self.body[0][1] * CELL
        pygame.draw.rect(screen, HEAD_COLOR, (head_x, head_y, CELL, CELL))
````

## 9.4. Файл `game.py` — игровое окно и цикл

Здесь располагается класс `Game` — он управляет запуском Pygame, читает нажатия и вызывает методы объекта-змейки. Строка `from snake import Snake` позволяет использовать класс из **другого файла**.

````python
%%writefile snake_oop/game.py
import pygame
from settings import WIDTH, HEIGHT, BACKGROUND, SPEED
from snake import Snake


class Game:
    def __init__(self):
        pygame.init()
        self.screen = pygame.display.set_mode((WIDTH, HEIGHT))
        pygame.display.set_caption("Змейка — урок 1")
        self.clock = pygame.time.Clock()
        self.snake = Snake()

    def run(self):
        running = True
        while running:
            # Читаем события мыши, окна и клавиатуры
            for event in pygame.event.get():
                if event.type == pygame.QUIT:
                    running = False
                elif event.type == pygame.KEYDOWN:
                    self.snake.turn(event.key)

            # Обновляем змейку и отображаем игру
            self.snake.move()
            if not self.snake.moving:
                pygame.display.set_caption("Змейка у стены! Закройте окно")

            self.screen.fill(BACKGROUND)
            self.snake.draw(self.screen)
            pygame.display.flip()
            self.clock.tick(SPEED)

        pygame.quit()
````

````python
%%writefile snake_oop/main.py
from game import Game


if __name__ == "__main__":
    game = Game()
    game.run()
````
